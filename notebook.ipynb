```python
from shop_assistant import (
    Document,
    Embedder,
    VectorStore,
    ContextBuilder,
    PromptBuilder,
    Generator,
)
import pandas as pd
```

```python
pd.set_option("display.max_colwidth", None)
```

```python
document = Document("./rules.txt")
embedder = Embedder()
```

```python
# 1. Load Documents
documents = document.load()
```

```python
# 2. Embed Documents
embeddings = embedder.embed(texts=documents)
```

```python
# 3. Store
store = VectorStore(documents=documents, embeddings=embeddings)
```

```python
# 4. Search
query = "تا چه زمانی میتونم سفارش رو مرجوع کنم؟"
query_embedding = embedder.embed(texts=[query])
retrieved_documents = store.search(query_embedding=query_embedding[0])

# pd.DataFrame(retrieved_documents)
```

```python
# 5. Generate Context
context_builder = ContextBuilder(retrieved_documents=retrieved_documents)
context = context_builder.build()
# print(context)
```

```python
# 6. Generate Prompt
prompt_builder = PromptBuilder(query=query, context=context)

prompt = prompt_builder.build()

# print(prompt)
```

```python
# 7. Chat
generator = Generator()

answer = generator.answer(prompt=prompt)

print(answer)
```
